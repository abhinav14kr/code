Write a function that takes a NumPy array of numbers and normalizes it so that all values fall between 0 and 1.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
print(arr)
```

```python
min_val = np.min(arr)
max_val = np.max(arr)
scaled_data = (arr - min_val) / (max_val - min_val)
print(scaled_data)
```

Suppose you have a matrix representing units sold for different products across regions, and another matrix representing the price per unit for each product. Write a NumPy script to calculate the total revenue per region using matrix multiplication.

```python
import numpy as np
import pandas as pd
```

```python
# Units sold (rows = regions, cols = products)
units = np.array([
    [10, 20, 30],   # Region A
    [5,  15, 25],   # Region B
    [2,  8,  12]    # Region C
])

# Prices per product
prices = np.array([
    [100],
    [200],
    [300]
])

print(prices)
```

```python
output = np.matmul(units, prices)
print(output)
```

- Convert the result into a Pandas DataFrame with region names (["Region A", "Region B", "Region C"]).
- Add a Total Revenue column summing across all regions.
- Write a function calculate_revenue(units, prices) with type hints and docstring.

```python
df = pd.DataFrame(output, index = ['Region A', 'Region B', 'Region C'])
print(df)
```

```python
df['total_revenue'] = df.sum(axis=1)
print(df)
```

Write a Pandas script to pivot the data so that each region becomes a column, with sales values filled in per date.

```python
import pandas as pd
from io import StringIO
import excel 
```

```python
data = """Date,Region,Sales
2024-01-01,North,100
2024-01-01,South,200
2024-01-02,North,150
2024-01-02,South,250
"""

# Using StringIO to simulate a file-like object
df = pd.read_csv(StringIO(data))
print(df)
```

```python
df = df.pivot_table(values='Sales', index='Date', columns='Region', aggfunc='sum') # pivot table syntax 
print(df)
```

```python
# total column summing across all regions

df['Total'] = df.sum(axis=1)
print(df)
```

```python
# Handle missing values by filling them with 0.
df = df.fillna(0)
```

```python
# Save the result to a new Excel file.
df.to_excel('sales_data_pivot.xlsx')
```

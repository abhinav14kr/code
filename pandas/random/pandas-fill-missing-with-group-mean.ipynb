Write a Pandas script to fill missing salaries with the average salary of that employee’s department.

```python
import pandas as pd
from io import StringIO

data = """Department,Employee,Salary
IT,John,5000
IT,Alice,5500
HR,Bob,
HR,Eva,4500
IT,Mark
"""

# Using StringIO to simulate a file-like object
df = pd.read_csv(StringIO(data))


print(df)
```

```python
df['average_salary'] = df.groupby('Department')['Salary'].transform('mean').round(2)
df['Salary'] = df['Salary'].fillna(df['average_salary'])
print(df)
```

Write a Pandas script to find the top N products by total sales across all categories.

```python
import pandas as pd 
import numpy as np

df = pd.read_csv('1.csv')
print(df)
```

```python
N = 3

grouped = df.groupby(['Product', 'Category'], as_index=False)['Sales'].sum()

sorted_df = grouped.sort_values(by='Sales', ascending=False)
sorted_df = sorted_df.rename(columns={'Sales': 'Total_Sales'})

top_n = sorted_df.head(N)
print(top_n)
```

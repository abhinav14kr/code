Write a Pandas script to calculate the total revenue per month sorted by month in ascending order.

```python
import pandas as pd
from io import StringIO

data = """OrderID,OrderDate,Amount
A001,2024-01-15,250
A002,2024-01-20,100
A003,2024-02-10,300
A004,2024-01-25,50
A005,2024-02-15,200"""

# Using StringIO to simulate a file-like object
df = pd.read_csv(StringIO(data))

df['OrderDate'] = pd.to_datetime(df['OrderDate'])

print(df)
```

```python
monthly_revenue = df.groupby(df['OrderDate'].dt.to_period('M'))['Amount'].sum()
monthly_revenue
```

```python
monthly_revenue = monthly_revenue.sort_index()
print(monthly_revenue)
```
